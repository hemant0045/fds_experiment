```python
# Experiment 1: Data Preprocessing and Cleaning
# Dataset: Titanic train.csv
```

```python
import pandas as pd
import numpy as np
```

```python
# Step 1: Load the dataset
df = pd.read_csv("train.csv")

print("Original Dataset:")
print(df.head())
```

```python
# Step 2: Dataset information
print("\nDataset Shape:")
print(df.shape)

print("\nDataset Information:")
df.info()
```

```python
# Step 3: Check missing values
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
# Step 4: Calculate statistical measures
print("\nStatistical Description:")
print(df.describe())
```

```python
# Step 5: Handle missing numerical values
df['Age'] = df['Age'].fillna(df['Age'].median())
df['Fare'] = df['Fare'].fillna(df['Fare'].median())
```

```python
# Step 6: Handle missing categorical values
df['Embarked'] = df['Embarked'].fillna(
    df['Embarked'].mode()[0]
)
```

```python
# Step 7: Handle Cabin missing values
df['Cabin'] = df['Cabin'].fillna('Unknown')
```

```python
# Step 8: Remove duplicate records
df.drop_duplicates(inplace=True)
```

```python
# Step 9: Check missing values after cleaning
print("\nMissing Values After Cleaning:")
print(df.isnull().sum())
```

```python
# Step 10: Analyze correlation
print("\nCorrelation Matrix:")
print(df.select_dtypes(include=np.number).corr())
```

```python
# Step 11: Analyze survival statistics
print("\nSurvival Count:")
print(df['Survived'].value_counts())

print("\nSurvival Rate:")
print(df['Survived'].mean())
```

```python
# Step 12: Relationship between gender and survival
print("\nGender-wise Survival:")
print(pd.crosstab(df['Sex'], df['Survived']))
```

```python
# Step 13: Relationship between passenger class and survival
print("\nPassenger Class-wise Survival:")
print(pd.crosstab(df['Pclass'], df['Survived']))
```

```python
# Step 14: Display cleaned dataset
print("\nCleaned Dataset:")
print(df.head())
```

```python
# Step 15: Save cleaned dataset
df.to_csv("titanic_cleaned.csv", index=False)

print("\nCleaned dataset saved successfully!")
```
